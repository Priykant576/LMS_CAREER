# #HEALTH DATA ANALYSIS

```python
import pandas as pd
import numpy as np

# HEALTH DATA ANALYSIS PROJECT

# 1. Create Dataset
data = {
    'Date': [
        '2026-01-01', '2026-01-03', '2026-01-05',
        '2026-01-08', '2026-01-10', '2026-02-02',
        '2026-02-05', '2026-02-08', '2026-02-12',
        '2026-02-15', '2026-03-01', '2026-03-04',
        '2026-03-08', '2026-03-12', '2026-03-15'
    ],
    'Patient_ID': [
        101, 102, 103, 104, 105,
        101, 106, 107, 108, 102,
        109, 110, 103, 111, 105
    ],
    'Age': [22, 35, 45, 28, 50, 22, 60, 32, 40, 35, 55, 25, 45, 38, 50],
    'Gender': [
        'Male', 'Female', 'Male', 'Female', 'Male',
        'Male', 'Female', 'Male', 'Female', 'Female',
        'Male', 'Female', 'Male', 'Female', 'Male'
    ],
    'Department': [
        'Cardiology', 'General', 'Orthopedic', 'Dental', 'Cardiology',
        'General', 'Cardiology', 'Dental', 'Orthopedic', 'General',
        'Cardiology', 'Dental', 'Orthopedic', 'General', 'Cardiology'
    ],
    'Branch': [
        'Delhi', 'Meerut', 'Delhi', 'Noida', 'Meerut',
        'Delhi', 'Noida', 'Meerut', 'Delhi', 'Noida',
        'Meerut', 'Delhi', 'Noida', 'Meerut', 'Delhi'
    ],
    'Treatment_Cost': [
        5000, 1500, 4000, 1000, 6000,
        2000, 7000, 1200, 3500, 1800,
        5500, 1000, 4200, 1600, 6500
    ],
    'Discount': [
        500, 100, 400, 50, 600,
        200, 700, 100, 300, 150,
        500, 50, 400, 100, 600
    ],
    'Blood_Pressure': [
        120, 118, 145, 110, 150,
        122, 160, 115, 135, 119,
        155, 112, 140, 117, 148
    ],
    'Heart_Rate': [
        72, 75, 90, 68, 95,
        74, 100, 70, 88, 76,
        96, 69, 85, 73, 92
    ],
    'Height': [
        1.70, 1.65, 1.75, 1.60, 1.80,
        1.70, 1.62, 1.72, 1.68, 1.65,
        1.78, 1.58, 1.75, 1.63, 1.80
    ],
    'Weight': [
        65, 55, 85, 50, 90,
        66, 80, 62, 75, 56,
        88, 48, 82, 58, 92
    ]
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'])

# Calculate BMI using NumPy
df['BMI'] = np.round(
    df['Weight'] / (df['Height'] ** 2), 2
)

df['Final_Cost'] = df['Treatment_Cost'] - df['Discount']

df['Month'] = df['Date'].dt.month_name()

df['Weekday'] = df['Date'].dt.day_name()


# QUESTION 1:
# What is the overall healthcare cost and how does it change over time?

print("\n1. OVERALL HEALTHCARE COST")

print("Total Treatment Cost:",
      df['Treatment_Cost'].sum())

print("Total Discount:",
      df['Discount'].sum())

print("Total Final Cost:",
      df['Final_Cost'].sum())

print("\nMonthly Healthcare Cost:")
print(df.groupby('Month')['Final_Cost'].sum())


# QUESTION 2:
# Which departments contribute the largest and smallest share?

print("\n2. DEPARTMENT-WISE ANALYSIS")

department_cost = df.groupby('Department')['Final_Cost'].sum()

print(department_cost.sort_values(ascending=False))

department_share = (
    department_cost / department_cost.sum()
) * 100

print("\nDepartment Revenue Share (%):")
print(np.round(department_share, 2))


# QUESTION 3:
# Which branches perform best?

print("\n3. BRANCH-WISE ANALYSIS")

print(df.groupby('Branch').agg(
    Total_Patients=('Patient_ID', 'count'),
    Total_Cost=('Final_Cost', 'sum'),
    Average_Cost=('Final_Cost', 'mean')
).sort_values(by='Total_Cost', ascending=False))


# QUESTION 4:
# Which months and weekdays show more patient visits?

print("\n4. MONTH AND WEEKDAY ANALYSIS")

print("\nMonthly Patient Visits:")
print(df.groupby('Month')['Patient_ID'].count())

print("\nWeekday Patient Visits:")
print(df.groupby('Weekday')['Patient_ID'].count())


# QUESTION 5:
# How does average treatment cost vary across departments and branches?

print("\n5. AVERAGE TREATMENT COST")

print("\nDepartment-wise Average Cost:")
print(df.groupby('Department')['Final_Cost'].mean())

print("\nBranch-wise Average Cost:")
print(df.groupby('Branch')['Final_Cost'].mean())


# QUESTION 6:
# Are larger discounts associated with higher treatment costs?

print("\n6. DISCOUNT ANALYSIS")

print("Average Discount:",
      np.mean(df['Discount']))

print("Average Treatment Cost:",
      np.mean(df['Treatment_Cost']))

correlation = np.corrcoef(
    df['Discount'], df['Treatment_Cost']
)[0, 1]

print("Discount and Treatment Cost Correlation:",
      round(correlation, 2))


# QUESTION 7:
# Which patients make frequent or high-value visits?

print("\n7. PATIENT-WISE ANALYSIS")

patient_analysis = df.groupby('Patient_ID').agg(
    Total_Visits=('Patient_ID', 'count'),
    Total_Expenses=('Final_Cost', 'sum'),
    Average_Expenses=('Final_Cost', 'mean')
)

print(patient_analysis.sort_values(
    by='Total_Expenses', ascending=False
))


# QUESTION 8:
# What actions could improve healthcare services?

print("\n8. HEALTHCARE IMPROVEMENT ANALYSIS")

print("Average Patient Age:", np.mean(df['Age']))

print("Average BMI:", round(np.mean(df['BMI']), 2))

print("Average Blood Pressure:",
      np.mean(df['Blood_Pressure']))

print("Average Heart Rate:",
      np.mean(df['Heart_Rate']))

print("\nPatients with High BMI:")
print(df[df['BMI'] >= 30][['Patient_ID', 'BMI']])

print("\nPatients with High Blood Pressure:")
print(df[df['Blood_Pressure'] >= 140][
    ['Patient_ID', 'Blood_Pressure']
])

print("\nDepartment Needing More Attention:")
print(department_cost.idxmin())

print("\nBranch with Highest Healthcare Cost:")
print(df.groupby('Branch')['Final_Cost'].sum().idxmax())


# FINAL DATA SUMMARY

print("\n FINAL HEALTH DATA SUMMARY")

print(df)

print("\nStatistical Summary:")
print(df.describe())

print("\nTotal Number of Records:", len(df))

print("Total Unique Patients:",
      df['Patient_ID'].nunique())

```
